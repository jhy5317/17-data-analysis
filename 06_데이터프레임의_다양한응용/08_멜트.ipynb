```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["John", "Mary"],
    "City": ["New York", "Los Angeles"],
    "Age": [28, 32],
    "Salary": [50000, 60000]
})

df
```

```python
df_melted = df.melt(id_vars=["Name", "City"],
                    var_name="Attribute",
                    value_name="Value")

df_melted
```
