```python
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib import font_manager, rc
font_path = "./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc("font", family=font_name)
```

```python
df = pd.read_excel("./data/시도별_전출입_인구수.xlsx")

df = df.ffill()
df.head()
```

```python
mask = (df["전출지별"] == "서울특별시") & (df["전입지별"] != "서울특별시")
df_seoul = df[mask]
df_seoul = df_seoul.drop(["전출지별"], axis=1)
df_seoul = df_seoul.rename({"전입지별":"전입지"}, axis=1)
df_seoul = df_seoul.set_index("전입지")
df_seoul
```

```python
sr_one = df_seoul.loc["경기도"]
sr_one.head()
```

```python
plt.plot(sr_one.index, sr_one.values)
```

```python
plt.plot(sr_one)
plt.show()
```

```python
## 차트 제목, 축 이름 추가, 선 스타일 지정

plt.plot(sr_one.index, sr_one.values, linestyle="--")

plt.title("서울 -> 경기 인구 이동")

plt.xlabel("기간")
plt.ylabel("이동 인구수")

plt.show()
```

```python
sr_one
```

```python
plt.style.use("ggplot")

plt.figure(figsize=(14,5))

plt.xticks(size=10, rotation='vertical')

plt.plot(
    sr_one.index, sr_one.values,
    marker="o",
    markerfacecolor="red",
    markeredgecolor="blue",
    markeredgewidth=2,
    markersize=10
)

plt.title("서울 -> 경기 인구 이동", size=30)
plt.xlabel("기간", size=20)
plt.ylabel("이동 인구수", size=20)

plt.legend(labels=["서울 -> 경기"], loc="best", fontsize=15)

plt.show()
```

```python
 ### 퀴즈 ###

# 1. 대구 -> 서울

df.head()
mask1 = (df["전출지별"] == "대구광역시")
df_daegu = df[mask1]

df_daegutoseoul = df_daegu.drop(["전출지별"], axis=1)
df_daegutoseoul = df_daegutoseoul.set_index("전입지별")
df_daegutoseoul

sr_one = df_daegutoseoul.loc["서울특별시"]

sr_one = sr_one.replace("-", pd.NA)
sr_one = sr_one.dropna()

print(sr_one)

plt.style.use("ggplot")

plt.figure(figsize=(14,5))

plt.xticks(size=10, rotation='vertical')

plt.plot(
    sr_one.index, sr_one.values,
    marker="o",
    markerfacecolor="red",
    markeredgecolor="blue",
    markeredgewidth=2,
    markersize=10
)

plt.title("대구 -> 서울 인구 이동", size=30)
plt.xlabel("기간", size=20)
plt.ylabel("이동 인구수", size=20)

plt.legend(labels=["대구 -> 서울"], loc="best", fontsize=15)

plt.show()

# 2. 경북 -> 대구

mask1 = (df["전출지별"] == "경상북도")
df_daegu = df[mask1]

df_daegutoseoul = df_daegu.drop(["전출지별"], axis=1)
df_daegutoseoul = df_daegutoseoul.set_index("전입지별")
df_daegutoseoul

sr_two = df_daegutoseoul.loc["대구광역시"]

sr_two = sr_two.replace("-", pd.NA)
sr_two = sr_two.dropna()

print(sr_two)

plt.style.use("ggplot")

plt.figure(figsize=(14,5))

plt.xticks(size=10, rotation='vertical')

plt.plot(
    sr_two.index, sr_two.values,
    marker="o",
    markerfacecolor="red",
    markeredgecolor="blue",
    markeredgewidth=2,
    markersize=10
)

plt.title("경북 -> 대구 인구 이동", size=30)
plt.xlabel("기간", size=20)
plt.ylabel("이동 인구수", size=20)

plt.legend(labels=["경북 -> 대구"], loc="best", fontsize=15)

plt.show()
```
